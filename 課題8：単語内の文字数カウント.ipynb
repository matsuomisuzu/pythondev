```python
#英単語を格納するための空のリスト
word = []

#辞書
a = {'a':0, 'b':0, 'c':0, 'd':0, 'e':0, 'f':0, 'g':0, 'h':0, 'i':0, 'j':0, 'k':0, 'l':0, 'm':0, 'n':0, 'o':0, 'p':0, 'q':0, 'r':0, 's':0, 't':0, 'u':0, 'v':0, 'w':0, 'x':0, 'y':0, 'z':0}


# 英単語の入力を繰り返す
while True:
    text = input("英単語を入力してください:")

    # 何も入力せずにEnterを押したら終了
    if text == "":
        break

    # 入力された英単語をリストに追加
    word.append(text)

    # 入力された英単語の文字を1文字ずつ数える
    for b in text:
        a[b] += 1

# 入力された英単語をABC順に並べ替える
word.sort()

# 入力された英単語のリストを表示
print(f"入力した英単語：{word}")

# 1文字以上あった文字だけ表示
for key, value in a.items():
    if value > 0:
        print(f"{key}が{value}個ありました")

```
